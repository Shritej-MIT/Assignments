```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Take CSV file name as input from the user
file_name = input("Enter the CSV file name: ")

# Load the CSV dataset
df = pd.read_csv(file_name, skipinitialspace=True)

# Display the dataset
print("\nDataset:")
print(df)

# Calculate and display averages
print("\nAverage Marks and Attendance:")
print("Average Maths Marks:", df["Maths"].mean())
print("Average Python Marks:", df["Python"].mean())
print("Average Attendance:", df["Attendance"].mean())


# 1. Line Plot
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o", label="Python Marks")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# 2. Bar Plot
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Maths"], label="Maths Marks")
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.legend()
plt.tight_layout()
plt.show()


# 3. Scatter Plot
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Maths"], df["Python"], s=70, label="Students")
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# 4. Correlation Matrix and Heatmap
correlation = df[["Maths", "Python", "Attendance"]].corr()

print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(6.5, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues",
    vmin=-1,
    vmax=1
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()


# 5. Histogram of Python Marks
plt.figure(figsize=(7, 4.5))
plt.hist(df["Python"], bins=5, edgecolor="black", label="Python Marks")
plt.title("Distribution of Python Marks")
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")
plt.legend()
plt.grid(axis="y")
plt.tight_layout()
plt.show()


# 6. Box Plot for Maths and Python Marks
plt.figure(figsize=(7, 4.5))
plt.boxplot(
    [df["Maths"], df["Python"]],
    tick_labels=["Maths", "Python"]
)
plt.title("Comparison of Maths and Python Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.tight_layout()
plt.show()


# 7. Pie Chart Showing Students from Each City
city_counts = df["City"].value_counts()

plt.figure(figsize=(7, 5))
plt.pie(
    city_counts,
    labels=city_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Percentage of Students from Each City")
plt.legend(city_counts.index, title="Cities")
plt.tight_layout()
plt.show()

```
